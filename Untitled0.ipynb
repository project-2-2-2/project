```python

import IPython.display as ipd
import torchaudio
import torch
from audiocraft.models import MusicGen
import ipywidgets as widgets
import random

# Load the MusicGen model
model = MusicGen.get_pretrained('facebook/musicgen-small')
model.set_generation_params(duration=10)  # Generate 10 seconds of music

# Dropdown for mood selection
mood = widgets.Dropdown(
    options=['Happy', 'Sad', 'Intense', 'Scary', 'Emotional'],
    description='Mood:',
    style={'description_width': 'initial'}
)
display(mood)

# Mapping moods to prompts
mood_prompts = {
    'Happy': 'A cheerful, uplifting melody with bright instruments.',
    'Sad': 'A slow, melancholic tune with emotional depth.',
    'Intense': 'A fast-paced, action-packed music with deep bass.',
    'Scary': 'A haunting, eerie atmosphere with unsettling tones.',
    'Emotional': 'A touching, deeply moving orchestral piece.'
}

```

```python
def generate_music(_):
    prompt = mood_prompts.get(mood.value, 'A generic instrumental piece.')
    print(f"Generating music for mood: {mood.value}")

    # Generate music
    wav = model.generate([prompt], progress=True)

    # Check if audio is generated
    if wav is None or len(wav) == 0:
        print("Error: No audio generated!")
        return

    # Save and play the generated audio
    file_path = 'generated_music.wav'
    torchaudio.save(file_path, wav[0].cpu(), 32000)

    print("Music generated successfully!")
    display(ipd.Audio(file_path))
generate_btn = widgets.Button(description='Generate Music')
generate_btn.on_click(generate_music)
display(generate_btn)
```
